```sql
USE CATALOG rearc;
USE SCHEMA silver;

CREATE OR REPLACE TABLE productivity AS

SELECT
    p.series_id,
    p.year,
    p.period,
    per.period_name,
    CAST(p.value AS DOUBLE) AS value,

    s.sector_name,
    m.measure_text,
    c.class_text,

    ser.seasonal,
    ser.base_year

FROM rearc.bronze.productivity_data p

JOIN rearc.bronze.series ser
ON p.series_id = ser.series_id

LEFT JOIN rearc.bronze.sector s
ON ser.sector_code = s.sector_code

LEFT JOIN rearc.bronze.measure m
ON ser.measure_code = m.measure_code

LEFT JOIN rearc.bronze.class_lookup c
ON ser.class_code = c.class_code

LEFT JOIN rearc.bronze.period per
ON p.period = per.period;
```

```sql
SELECT *
FROM productivity
LIMIT 20;
```

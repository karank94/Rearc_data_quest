```python
import requests
from bs4 import BeautifulSoup

BLS_URL = "https://download.bls.gov/pub/time.series/pr/"

HEADERS = {
    "User-Agent": "Rearc Data Quest (Karan Kaushal; karankaushal94@gmail.com)"
}

POPULATION_API_URL = (
    "https://honolulu-api.datausa.io/tesseract/data.jsonrecords"
    "?cube=acs_yg_total_population_1"
    "&drilldowns=Year,Nation"
    "&locale=en"
    "&measures=Population"
)

LANDING_PATH = "/Volumes/rearc/raw/landing"
```

```python
def discover_bls_files(bls_url):
    response = requests.get(bls_url, headers=HEADERS)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    files = []

    for link in soup.find_all("a"):
        href = link.get("href")

        if href and "pr." in href:
            files.append(href.split("/")[-1])

    return files
```

```python
bls_files = discover_bls_files(BLS_URL)

print(bls_files)
```

```python
def download_file(filename):
    """
    Downloads a BLS file and stores it in the landing volume.
    """

    file_url = BLS_URL + filename

    response = requests.get(file_url, headers=HEADERS)
    response.raise_for_status()

    output_path = f"{LANDING_PATH}/{filename}"

    with open(output_path, "wb") as f:
        f.write(response.content)

    print(f"Downloaded {filename}")
```

```python
for filename in bls_files:
    download_file(filename)

print("All BLS files downloaded successfully!")
```

```python

def download_population_data():

    response = requests.get(
        POPULATION_API_URL,
        headers=HEADERS
    )

    response.raise_for_status()
    output_path = f"{LANDING_PATH}/population.json"

    with open(output_path, "w") as f:
        f.write(response.text)

    print("Downloaded population.json")
```

```python
download_population_data()
```
